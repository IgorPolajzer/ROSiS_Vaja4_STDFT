1. Kako je nestacionarnost signalov odvisna od hitrosti izgovarjave oz. ali se optimalna izbira dolžine intervalov za DFT in stopnje njihovega prekrivanja razlikuje od hitrosti izgovarjave?

O: Hitrejša je izgovorjava, bolj nestacionaren je signal, saj se frekvence hitreje spreminjajo. Pri hitri izgovorjavi so zato potrebni krajši časovni intervali z večjim prekrianjem, da ohranimo zadostno časovno ločljivost.

```python
TEST_FILES = ['a_medium_pitch.mp3', 'i_medium_pitch.mp3', 'o_medium_pitch.mp3']

for file in TEST_FILES:
    for interval in INTERVAL_LENGTHS:
            fr, ti, sp = stdft(f"/home/igor/Desktop/MAG/1_LETNIK/2_SEMESTER/RACUNALNISKA_OBDELAVA_SIGNALOV_IN_SLIK/Vaja_4/recordings/{file}", interval, 0.0)
            plot_spectogram(fr, ti, sp, file, interval, overlap, True)
```
